```python
import csv
import os
import shutil

# Original dataset
original_file = "college_student_placement_dataset.csv"

# Working copy for CRUD operations
filename = "college_student_placement_crud.csv"

# Create working copy only if it does not already exist
if not os.path.exists(filename):
    shutil.copy(original_file, filename)
    print("Working copy created successfully.")
else:
    print("Working copy already exists.")


# Column names of the placement dataset
fields = [
    "College_ID",
    "IQ",
    "Prev_Sem_Result",
    "CGPA",
    "Academic_Performance",
    "Internship_Experience",
    "Extra_Curricular_Score",
    "Communication_Skills",
    "Projects_Completed",
    "Placement"
]

#---------------- CREATE ----------------

def add_student():

    print("\n--- Add New Student ---")

    student = [
        input("College ID: "),
        input("IQ: "),
        input("Previous Semester Result: "),
        input("CGPA: "),
        input("Academic Performance: "),
        input("Internship Experience (Yes/No): "),
        input("Extra Curricular Score: "),
        input("Communication Skills: "),
        input("Projects Completed: "),
        input("Placement (Yes/No): ")
    ]

    with open(filename, "a", newline="") as file:

        writer = csv.writer(file)

        writer.writerow(student)

    print("Student added successfully.")


# ---------------- READ ----------------

def display_students():

    print("\n--- Student Records ---")

    try:

        with open(filename, "r", newline="") as file:

            reader = csv.DictReader(file)

            for student in reader:

                print(
                    student["College_ID"],
                    "| IQ:", student["IQ"],
                    "| Prev Result:", student["Prev_Sem_Result"],
                    "| CGPA:", student["CGPA"],
                    "| Academic:", student["Academic_Performance"],
                    "| Internship:", student["Internship_Experience"],
                    "| Extra Curricular:", student["Extra_Curricular_Score"],
                    "| Communication:", student["Communication_Skills"],
                    "| Projects:", student["Projects_Completed"],
                    "| Placement:", student["Placement"]
                )

    except FileNotFoundError:

        print("CSV file not found.")


# ---------------- SEARCH ----------------

def search_student():

    print("\n--- Search Student ---")

    college_id = input("Enter College ID: ")

    found = False

    try:

        with open(filename, "r", newline="") as file:

            reader = csv.DictReader(file)

            for student in reader:

                if student["College_ID"] == college_id:

                    print("\nStudent Found")
                    print("----------------------")

                    for field in fields:
                        print(field, ":", student[field])

                    found = True
                    break

        if not found:
            print("Student not found.")

    except FileNotFoundError:

        print("CSV file not found.")


# ---------------- UPDATE ----------------

def update_student():

    print("\n--- Update Student ---")

    college_id = input("Enter College ID to update: ")

    students = []
    found = False

    try:

        with open(filename, "r", newline="") as file:

            reader = csv.DictReader(file)

            for student in reader:

                if student["College_ID"] == college_id:

                    found = True

                    print("\nEnter new details:")

                    student["IQ"] = input("IQ: ")
                    student["Prev_Sem_Result"] = input(
                        "Previous Semester Result: "
                    )
                    student["CGPA"] = input("CGPA: ")
                    student["Academic_Performance"] = input(
                        "Academic Performance: "
                    )
                    student["Internship_Experience"] = input(
                        "Internship Experience (Yes/No): "
                    )
                    student["Extra_Curricular_Score"] = input(
                        "Extra Curricular Score: "
                    )
                    student["Communication_Skills"] = input(
                        "Communication Skills: "
                    )
                    student["Projects_Completed"] = input(
                        "Projects Completed: "
                    )
                    student["Placement"] = input(
                        "Placement (Yes/No): "
                    )

                students.append(student)


        if found:

            with open(filename, "w", newline="") as file:

                writer = csv.DictWriter(
                    file,
                    fieldnames=fields
                )

                writer.writeheader()

                writer.writerows(students)

            print("Student record updated successfully.")

        else:

            print("Student not found.")

    except FileNotFoundError:

        print("CSV file not found.")


# ---------------- DELETE ----------------

def delete_student():

    print("\n--- Delete Student ---")

    college_id = input("Enter College ID to delete: ")

    students = []
    found = False

    try:

        with open(filename, "r", newline="") as file:

            reader = csv.DictReader(file)

            for student in reader:

                if student["College_ID"] == college_id:

                    found = True

                else:

                    students.append(student)


        if found:

            with open(filename, "w", newline="") as file:

                writer = csv.DictWriter(
                    file,
                    fieldnames=fields
                )

                writer.writeheader()

                writer.writerows(students)

            print("Student deleted successfully.")

        else:

            print("Student not found.")

    except FileNotFoundError:

        print("CSV file not found.")


# ---------------- MENU ----------------

while True:

    print("\n===================================")
    print(" Student Placement CSV Management")
    print("===================================")

    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":

        add_student()

    elif choice == "2":

        display_students()

    elif choice == "3":

        search_student()

    elif choice == "4":

        update_student()

    elif choice == "5":

        delete_student()

    elif choice == "6":

        print("Program ended.")
        break

    else:

        print("Invalid choice. Please try again.")
```

```python
academic_percentage = (
    data["Academic_Performance"] / 10
) * 100

print(academic_percentage.head())

t_statistic, p_value = stats.ttest_1samp(
    academic_percentage,
    40
)

print("T-Statistic:", t_statistic)
print("P-Value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
    print("The average academic percentage is significantly different from 40.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is not enough evidence to conclude that the average academic percentage is different from 40.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

ml_data = data.copy()

ml_data["Internship_Experience"] = ml_data[
    "Internship_Experience"
].map({
    "Yes": 1,
    "No": 0
})

ml_data["Placement"] = ml_data[
    "Placement"
].map({
    "Yes": 1,
    "No": 0
})
```

```python
features = [
    "IQ",
    "Prev_Sem_Result",
    "CGPA",
    "Academic_Performance",
    "Internship_Experience",
    "Extra_Curricular_Score",
    "Communication_Skills",
    "Projects_Completed"
]

X = ml_data[features]
y = ml_data["Placement"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Model trained successfully!")

y_pred = model.predict(X_test)

print("Predictions:")
print(y_pred)

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100)

print(
    classification_report(
        y_test,
        y_pred
    )
)

cm = confusion_matrix(
    y_test,
    y_pred
)

print("Confusion Matrix:")
print(cm)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["Not Placed", "Placed"],
    yticklabels=["Not Placed", "Placed"]
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
feature_importance = pd.DataFrame({
    "Feature": features,
    "Importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature"
)

plt.title("Feature Importance in Student Placement Prediction")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()
```

```python
def cgpa_range(cgpa):
    if 4 <= cgpa < 6:
        return "4-6"
    elif 6 <= cgpa < 8:
        return "6-8"
    elif 8 <= cgpa <= 10:
        return "8-10"
    else:
        return "Outside Range"
```

```python
data["CGPA_Range"] = data["CGPA"].apply(cgpa_range)

placed_data = data[data["Placement"] == "Yes"]

cgpa_count = placed_data["CGPA_Range"].value_counts().reset_index()
cgpa_count.columns = ["CGPA_Range", "Count"]

# Exclude CGPA > 10
cgpa_count = cgpa_count[cgpa_count["CGPA_Range"] != "Outside Range"]

fig = px.pie(
    cgpa_count,
    names="CGPA_Range",
    values="Count",
    title="Placed Students by CGPA Range"
)

fig.show()
```
